# <font color='#E8800A'>Is this machine learning?</font> <a class="anchor" id="is-this-ml"></a>

Not every problem with data in it is a machine-learning problem, and
knowing the difference stops you building the wrong thing. Supervised machine
learning fits when three conditions hold at the same time:

1. a **pattern** plausibly connects what you know to what you want to predict;
2. you **cannot write the rule down** yourself, or the rule keeps changing;
3. you have **past examples with the answer attached** to learn from.

# <font color='#E8800A'>Supervised or unsupervised?</font> <a class="anchor" id="supervised"></a>
[Back to TOC](#toc)

Machine learning splits in two, and what separates the halves is
whether your past examples come with the answer attached.

- **Supervised**: every example carries the answer you will later want to
  predict: plantations already marked as drug plantations or legal crops, loans
  already marked repaid or defaulted. You learn the link between what you know
  and the answer, then apply that link to examples where the answer is still
  missing.
- **Unsupervised**: there is no answer column. You look for structure instead:
  which customers behave alike, which of forty measurements keep saying the same
  thing. Nobody can tell you the grouping you found is *correct*, because there
  is nothing to compare it against.

This course is about the supervised half, and the reason is checking. When the
answer is known for past examples, you can measure how often you would have been
right. Without answers, you can only look at the result and judge whether it
seems reasonable.

# <font color='#E8800A'>The five objects of a supervised problem</font> <a class="anchor" id="five-objects"></a>
[Back to TOC](#toc)

Here is the problem you will work on later in this notebook. An
environmental inspection agency photographs plantations from drones. Each
photograph yields four numeric measurements, named `BD1` to `BD4`. For 300
plantations an inspector has already visited the site and recorded whether it
was a **drug plantation** or a **legal crop**. Your boss now has 40 more
plantations that have measurements but no visit, and asks which of them to
inspect first.

That problem, and every other supervised problem, is built from the same five
objects. Learn to name them and half the vocabulary of the course is yours:

1. the **observation**, the thing you predict about: one plantation;
2. the **features**, what you know at the moment of predicting: `BD1` to `BD4`;
3. the **target**, what you want to know: drug plantation or legal crop;
4. the **prediction moment**, when the answer is needed, which fixes what
   information you are allowed to use: before anyone visits the site;
5. the **decision**, what happens as a result: an inspection team is sent, or
   is not.

The most common framing mistake is using information that would not exist yet at
the prediction moment. A model given the inspector's report as a feature would
score close to perfect and be useless, because a new plantation has no report.
That mistake has a name, **leakage**, and it returns later in the course.

Name the five objects for this problem: a bank wants to know, at the
moment a customer applies for a loan, whether that loan will be repaid, so that
it can approve or refuse the application.

```python
# Observation: one loan application / applicant.
# Features: information available when the customer applies, such as income,
#            requested amount, employment history, and credit history.
# Target: whether the loan will be repaid (repaid or defaulted).
# Prediction moment: when the application is submitted, before repayment is known.
# Decision: approve or refuse the loan application.
```

# <font color='#E8800A'>Where do the labels come from?</font> <a class="anchor" id="labels"></a>
[Back to TOC](#toc)

`BD1` to `BD4` came off an instrument. The drug-plantation label did
not. It is somebody's decision, made on a site visit, on whatever evidence was
in front of them that day, and typed into a file before you ever opened it.

Supervised learning copies that decision. Whatever is in the labels, including
the parts that are wrong, is what the model is being asked to reproduce. So
*"who produced this column, and how?"* is a question to ask before trusting
anything a model built from it tells you.

Three questions about the label column:

- **(a)** Name two ways a plantation in that file could end up with the wrong
  label.
- **(b)** Suppose every plantation smaller than one hectare had been labelled
  "legal crop" without anyone visiting it. What would a model learn from the
  file, and would its score reveal the problem?
- **(c)** What would you want to know about how that column was produced before
  letting a model act on new plantations?

```python
# (a) A visit or measurement may have been recorded incorrectly; the label
#     may also be mistyped, or the inspector may have misclassified the site.
# (b) The model would learn the shortcut "small means legal" rather than the
#     underlying plantation type. Its score might look good if the shortcut
#     agrees with the labels, so accuracy on the same file would not reveal
#     the labelling rule or its bias.
# (c) I would ask who made the decision, what evidence and threshold they used,
#     whether labels were independently checked, and how errors or ambiguous
#     cases were handled. I would also ask whether the labelling process changed
#     over time or differed between inspectors.
```

# <font color='#E8800A'>Classifying plantations by hand</font> <a class="anchor" id="by-hand"></a>

Now open the file. It sits in this repository at
`data/raw/drug_plantations.xlsx` and has two sheets: `ClassifiedData`, the 300
plantations whose answer is known, and `Data2Classify`, the 40 your boss is
asking about.

The cells that follow are provided. Run each one with **Shift+Enter** and read
what appears underneath it. None of them ends in `print(...)`: a notebook

Run this notebook from its own folder inside the course

```python
# Provided cell: run it (Shift+Enter). pandas is the library the course
# uses for tables; read_excel opens one sheet of a spreadsheet.
import pandas as pd

PLANTS = "../../data/raw/drug_plantations.xlsx"
plants_truth = pd.read_excel(PLANTS, sheet_name="ClassifiedData", index_col="ID")

plants_truth.head(8)
```

Eight rows out of 300. Four measurements, and one column `DrugPlant`
holding the answer: 1 for a drug plantation, 0 for a legal crop.

How many of each are there?

```python
# Provided cell: value_counts() counts how often each value appears.
plants_truth["DrugPlant"].value_counts().sort_index()
```

The two groups are nowhere near equal: there are roughly three legal
crops for every drug plantation. Counting the categories is one of the first
things to do when you meet a dataset, and an unequal count is a warning rather
than a detail. Three reasons to be wary of it:

- a model can score well by leaning on the larger group and never find a single
  case of the smaller one, which is normally the case the work exists to catch;
- the smaller group brings fewer examples to learn from, so whatever pattern it
  has is harder to pick up;
- any rule you write will look better than it is if you judge it on how often it
  agrees overall.

Keep the count in view. The closing section of this notebook is built on it.

Next, the table to reason from: the average of each measurement, taken
separately for the legal crops and for the drug plantations.

```python
# Provided cell: group the rows by the answer, then average each
# measurement within each group.
plants_truth.groupby("DrugPlant")[["BD1", "BD2", "BD3", "BD4"]].mean().round(1)
```

And the 40 plantations with no answer yet, the ones your boss wants
ranked.

```python
# Provided cell: the second sheet of the same file.
plants_2classify = pd.read_excel(PLANTS, sheet_name="Data2Classify", index_col="ID")

plants_2classify.head(8)
```

Four cells, four things on screen. Once the syntax is familiar, the
two you actually reason from (how many of each, and the average of each
measurement) fit in a single table built in one cell. You are not expected to
write cells like this yet; it is here so that a denser cell later in the course
is not a surprise.

```python
# Provided cell: the counting and the averaging, in one table.
summary = plants_truth.groupby("DrugPlant")[["BD1", "BD2", "BD3", "BD4"]].mean().round(1)
summary.insert(0, "plantations", plants_truth["DrugPlant"].value_counts().sort_index())

summary
```

Now do the job by hand. Take two minutes and write down a rule of the
form "if BD1 is above … and BD2 is below …, then it is a drug plantation",
reading your thresholds off the table of group averages. Then say how you would
find out whether your rule is any good, remembering that for 300 of these
plantations the answer is already known.

```python
# A simple rule read from the group averages is: classify as a drug
# plantation when BD1 > 24 and BD2 < 45; otherwise classify as legal.
# To assess it, apply it to labelled rows and compare predictions with truth,
# reporting accuracy and the confusion matrix (especially drug-plant recall).
# Because this uses the same rows that suggested the rule, also check it on
# held-out labelled rows or with cross-validation before using it on new sites.
truth = plants_truth["DrugPlant"]
my_rule = (plants_truth["BD1"] > 24) & (plants_truth["BD2"] < 45)
bd4_rule = plants_truth["BD4"] > 13

print("BD1 & BD2 rule: ", (truth == my_rule).mean() * 100, "%")
print("BD4 rule: ", (truth == bd4_rule).mean() * 100, "%")
```

<div class="alert alert-block alert-success">

**You have just done supervised machine learning by hand.** You used
examples whose answer was known (*training data*) to work out a rule (*a
model*), and you would apply that rule to plantations with no answer yet
(*prediction*). Everything ahead is a more careful and more automatic version of
what you just did, starting next class, on this same file, in Python.

# <font color='#E8800A'>What does "accurate" hide?</font> <a class="anchor" id="floor"></a>
[Back to TOC](#toc)

A number on its own is not a result. Before you can say whether 97 % is
good, you have to know what the same measurement gives for a model that does no
work at all.

Screening for a condition two people in a hundred have: a model that answers
**no** to everybody is right ninety-eight times out of a hundred. It finds
nothing, it cannot find anything, and it still beats most first attempts on
accuracy alone.

That do-nothing score has a name, the **floor**, and every result in this
course is reported against one.

The second provided cell counted how many of the 300 labelled
plantations are drug plantations and how many are legal crops.

- **(a)** What accuracy does a model that answers "legal crop" for every
  plantation get on those 300 rows?
- **(b)** A classmate reports 75 % accuracy and is pleased. What do you say?
- **(c)** Name one thing you would rather be told than accuracy, and say what it
  reveals that accuracy hides.

```python
# (a) It gets the proportion of labelled rows that are legal crops, which is
#     the majority-class baseline. It can be high without finding any drugs.
# (b) I would compare 75% with that baseline and inspect the confusion matrix.
#     If the baseline is close to 75%, the model has added little; it may also
#     be missing nearly every drug plantation despite its apparently good score.
# (c) I would rather see precision and recall (or an F1 score) for the drug
#     class, plus the confusion matrix. These reveal false negatives and false
#     positives, which overall accuracy hides.

baseline_accuracy = (truth == 0).mean()
print("Always-legal baseline accuracy:", baseline_accuracy * 100, "%")
```

# <font color='#E8800A'>Key takeaways</font> <a class="anchor" id="takeaways"></a>
[Back to TOC](#toc)

Six things to carry out of today:

1. **Supervised machine learning needs three conditions at once**: a plausible
   pattern, a rule you cannot write down yourself, and past examples with the
   answer attached. Missing one, use something simpler.
2. **Supervised and unsupervised differ by whether the answer is there.** With
   answers you can check yourself; without them you can only inspect the result.
3. **Five objects describe any supervised problem**: observation, features,
   target, prediction moment, decision. The prediction moment is what decides
   which features are legitimate.
4. **Labels are decisions, not measurements.** A model reproduces them, mistakes
   included, and no score computed on the same file can see it.
5. **Classification predicts a category; regression predicts a number.**
6. **A score means nothing on its own.** Report it against what a model that
   does no work would have got.
